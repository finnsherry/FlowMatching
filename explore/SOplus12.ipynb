```python
import torch
from lieflow.groups import SOplus12
```

```python
soplus12 = SOplus12()
```

```python
def random_time_like(N, ε_stab=1e-4):
    vec = torch.randn(N, 3)
    vec[..., 0] = (1 + torch.randn(N)**2).sqrt().clamp_min(1 + ε_stab) * (vec[..., 1]**2 + vec[..., 2]**2).sqrt()
    # restrict such that |angle| < pi
    return vec * ((1 - 1000. * ε_stab) * torch.pi / inner_product(vec, vec, keepdim=True).sqrt()).clamp_max(1.)

def random_space_like(N):
    vec = torch.randn(N, 3)
    vec[..., 0] = torch.clamp_min(1/(1 + torch.randn(N)**2).sqrt(), 0.1) * (vec[..., 1]**2 + vec[..., 2]**2).sqrt()
    return vec

def random_light_like(N):
    vec = torch.randn(N, 3)
    vec[..., 0] = (vec[..., 1]**2 + vec[..., 2]**2).sqrt()
    return vec

def inner_product(vec1, vec2, keepdim=False):
    return (vec1 @ soplus12.η * vec2).sum(-1, keepdim=keepdim)
```

```python
test_vec = random_time_like(4096)
test_gen = (test_vec[..., None, None] * soplus12.lie_algebra_basis).sum(-3)
(soplus12.exp(test_gen) - torch.matrix_exp(test_gen)).abs().max()
```

```python
(soplus12.log(soplus12.exp(test_gen)) - test_gen).abs().max()
```

```python
(soplus12.exp(soplus12.log(soplus12.exp(test_gen))) - soplus12.exp(test_gen)).abs().max()
```

```python
test_vec = random_space_like(4096)
test_gen = (test_vec[..., None, None] * soplus12.lie_algebra_basis).sum(-3)
(soplus12.exp(test_gen) - torch.matrix_exp(test_gen)).abs().max()
```

```python
(soplus12.exp(soplus12.log(soplus12.exp(test_gen))) - soplus12.exp(test_gen)).abs().max()
```

```python
test_vec = random_light_like(4096)
test_gen = (test_vec[..., None, None] * soplus12.lie_algebra_basis).sum(-3)
(soplus12.exp(test_gen) - torch.matrix_exp(test_gen)).abs().max()
```
